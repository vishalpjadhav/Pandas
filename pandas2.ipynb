```python
import pandas as pd

data = {
    'Name': ['Shivraj', 'Vaibhav', 'Vishal'],
    'Marks': [80, 90, 75]
}

df = pd.DataFrame(data)

# Insert new column using insert() function
df.insert(1, 'Age', [18, 19, 20])
df.insert(2,'Roll_no',[1008,1007,1004])
```

```python
# df["Marks"] = df["Marks"] 
df.loc[df["Marks"] > 80, "Marks"] = df["Marks"] + 5
```

# Handling Missing Data

```python
import pandas as pd

data = {
    'Name': ['Shivraj', 'Vaibhav', None],
    'Marks': [80, 90, None]
}

df = pd.DataFrame(data)
```

* Detect missing / NaN values using isnull()

```python
df.isnull().sum()
```

```python
df
```

* Fill missing values using fillna()

```python
df['Marks'] = df['Marks'].fillna(0)
df['Name'] = df['Name'].fillna("xyz")
```

```python
df
```

* Remove the Row where the value is missing 

```python
df.dropna(axis = 0 , inplace=True)
```

```python
df
```

```python
df.isnull()
```

* Fill Missing Values With mean 

```python
import pandas as pd

data = {
    'Name' : ['Vaibhav','Shivraj','Vishal','Ram'],
    'Marks': [80, None, 75, 90]
}

df = pd.DataFrame(data)

mean_value = df['Marks'].mean()

df['Marks'] = df['Marks'].fillna(mean_value)

print(df)
```
