# Handling Missing Data

```python
import pandas as pd

data = {
    'Name': ['Shivraj', 'Vaibhav', None],
    'Marks': [80, 90, None]
}

df = pd.DataFrame(data)
```

* Detect missing / NaN values using isnull()

```python
df.isnull().sum()
```

```python
df
```

* Fill missing values using fillna()

```python
df['Marks'] = df['Marks'].fillna(0)
df['Name'] = df['Name'].fillna("xyz")
```

```python
df
```

* Remove the Row where the value is missing 

```python
df.dropna(axis = 0 , inplace=True)
```

```python
df
```

```python
df.isnull()
```

* Fill Missing Values With mean 

```python
import pandas as pd

data = {
    'Name' : ['Vaibhav','Shivraj','Vishal','Ram'],
    'Marks': [80, None, 75, 90]
}

df = pd.DataFrame(data)

print(df)
print("After the change")
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

print(df)
```

```python
import pandas as pd 
data = {
    "Name" : ["Vishal","Shivraj","Vaibhav","Nagesh"],
    "Age" : [20,None,None,15]
}

df = pd.DataFrame(data)
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df
```

* Interpolation

```python
import pandas as pd

data = {
    'Name' : ['Vaibhav','Shivraj','Vishal','Ram'],
    'Marks': [80, None, 75, 90]
}

df = pd.DataFrame(data)
df['Marks']=df['Marks'].interpolate(method="linear",axis=0)
```

```python
df
```

* Methods in interpolation 
linear , polinomial , time 

```python
import pandas as pd

data = {
    'Marks': [80, None, 90, 100]
}

df = pd.DataFrame(data)

df = df.interpolate(method='linear')

print(df)
```

```python
import pandas as pd 
data = {
    "Time" : [1,2,3,4,5],
    "Value" : [10,None,30 ,None,50]
}
df = pd.DataFrame(data)
print("Before interpolation")
print(df)

df["Value"] = df["Value"].interpolate(method="linear")
print("After interpolation")
df
```

* When the interpolation used 
* 1 - Timer series data 
2 - numerical data with trends 
3 - Avoid dropping Rows 

# Sorting and aggregation

*Sorting data in one column
syntax :- df.sort_values(by="Column_name", True/False)

```python
import pandas as pd 
data = {
    "Name" :['Arun','Varun','Karun'],
    "Age" :[25,36,45],
    "salary":[25000,30000,4000]
}

df = pd.DataFrame(data)
#Sorted Age by the desceding order 
df.sort_values(by="Age",ascending=True,inplace=True)
```

```python
df
```

```python
# Sorted Age by asceding order 
df.sort_values(by="Age",ascending=False,inplace=True)
```

```python
df
```

* Sorting Values From Multiple Columns

* For multiple column sorting in Pandas We need to pass the ascending = [] , We Need to Pass True and False In List 

```python
df.sort_values(by=["Age","salary"], ascending=[True,True],inplace=True)
```

```python
df
```

# Aggregation in Pandas

* In pandas, aggregation means combining multiple values into a single summary value. It is commonly used when analyzing data to compute statistics like:

sum
mean (average)
count
min/max
median
standard deviation

Aggregation is often used with groupby().

```python
df["Age"].sum()
```

```python
df["Age"].mean()
```

```python
df["Age"].min()
```

```python
df["Age"].max()
```

```python
df["Age"].std()
```

```python
df["Age"].var()
```

# Grouping in Pandas

Grouping in pandas means dividing data into groups based on one or more columns so that you can perform calculations (sum, mean, count, etc.) on each group separately.

Pandas uses the groupby() function for grouping data.

```python
df.groupby('Department')['Salary'].sum()
```

```python
df
```

```python
df.groupby(['Department','Gender'])['Salary'].sum()
```
